Dieses Notebook demonstriert, wie Keras-Layer zur Data Augmentation genutzt werden können, um die Variabilität der Trainingsdaten zu erhöhen. Dies ist besonders nützlich, um Overfitting zu vermeiden.

```python
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
import numpy as np

# Fashion MNIST laden
fashion_mnist = keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

# Normalisierung und Kanal-Dimension
x_train = x_train.astype('float32') / 255.0
x_train = x_train[..., np.newaxis]

# Drei Beispielbilder auswählen
images = x_train[:3]

def plot_images(imgs, title):
    plt.figure(figsize=(10, 3))
    for i in range(len(imgs)):
        plt.subplot(1, 3, i + 1)
        plt.imshow(imgs[i].squeeze(), cmap='gray')
        plt.axis('off')
    plt.suptitle(title)
    plt.show()

plot_images(images, 'Originalbilder')
```

## RandomFlip
Dieser Layer spiegelt die Bilder horizontal oder vertikal.

```python
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal")
])

augmented_images = data_augmentation(images, training=True)
plot_images(augmented_images.numpy(), 'Nach RandomFlip (Horizontal)')
```

## RandomRotation
Dieser Layer rotiert die Bilder um einen zufälligen Winkel.

```python
data_augmentation = keras.Sequential([
    layers.RandomRotation(0.1)
])

augmented_images = data_augmentation(images, training=True)
plot_images(augmented_images.numpy(), 'Nach RandomRotation')
```

## RandomZoom
Dieser Layer zoomt zufällig in die Bilder hinein oder heraus.

```python
data_augmentation = keras.Sequential([
    layers.RandomZoom(0.3)
])

augmented_images = data_augmentation(images, training=True)
plot_images(augmented_images.numpy(), 'Nach RandomZoom')
```

## RandomTranslation
Dieser Layer verschiebt die Bilder zufällig horizontal und vertikal.

```python
data_augmentation = keras.Sequential([
    layers.RandomTranslation(height_factor=0.1, width_factor=0.1)
])

augmented_images = data_augmentation(images, training=True)
plot_images(augmented_images.numpy(), 'Nach RandomTranslation')
```

```python
data_augmentation = keras.Sequential([
    layers.RandomRotation(0.1),
    layers.RandomTranslation(height_factor=0.1, width_factor=0.1)
])

augmented_images = data_augmentation(images, training=True)
plot_images(augmented_images.numpy(), 'Nach RandomTranslation')
```
