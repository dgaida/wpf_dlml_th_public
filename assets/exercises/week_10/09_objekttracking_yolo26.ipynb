Wir laden ein Beispielvideo herunter, um das Tracking zu testen.

```python
import os

# Using a publicly accessible sample video from a different Google Cloud Storage bucket
video_url = "https://storage.googleapis.com/gweb-research-datasets-ai/simulated_traffic/simulated_traffic.mp4"
output_path = "/content/sample_video.mp4"

try:
    # Use wget to download the video from the provided URL
    !wget -O {output_path} {video_url}

    # Check if the file was actually downloaded successfully
    if os.path.exists(output_path) and os.path.getsize(output_path) > 0:
        print(f"Video erfolgreich heruntergeladen nach: {output_path}")
    else:
        print(f"Fehler: Video wurde nicht oder unvollständig heruntergeladen von {video_url}")
        # Optionally, raise an exception or handle this more robustly if needed
except Exception as e:
    print(f"Fehler beim Herunterladen des Videos: {e}")
```

## 3. Objekt Tracking  
  
Tracking verfolgt Objekte über mehrere Frames hinweg und weist ihnen eindeutige IDs zu.

```python
from ultralytics import YOLO

model = YOLO("yolo26n.pt")

# Tracking starten (BoT-SORT ist der Standard-Tracker)
# Wir speichern das Ergebnis unter /content/runs/detect/track/
results = model.track(source="/content/sample_video.mp4", show=False, save=True, name="track")
```

## 4. Video-Konvertierung  
  
Da YOLO Videos im `.avi`-Format speichert, konvertieren wir sie in `.mp4`, damit sie direkt im Notebook angezeigt werden können.

```python
# Installiere ffmpeg
!apt-get update -qq && apt-get install -y -qq ffmpeg

import os

# Pfade für das Video setzen
# Hinweis: Ultralytics speichert standardmäßig in runs/detect/track/ oder runs/detect/track2 etc.
input_avi_path = "/content/runs/detect/track/sample_video.avi"
output_mp4_path = "/content/runs/detect/track/sample_video_output.mp4"

# Konvertierung
if os.path.exists(input_avi_path):
    !ffmpeg -i {input_avi_path} -c:v libx264 -preset medium -crf 23 -y {output_mp4_path}
    print(f"Konvertierung abgeschlossen: {output_mp4_path}")
else:
    print(f"Datei {input_avi_path} nicht gefunden. Prüfe ggf. den Pfad in /content/runs/detect/.")
```

## 5. Ergebnisse anzeigen  
  
Nun zeigen wir das verarbeitete Video an.

```python
from IPython.display import Video
import os

if os.path.exists(output_mp4_path):
    display(Video(output_mp4_path, embed=True, width=640, height=360))
else:
    print("Video konnte nicht zur Anzeige gefunden werden.")
```
