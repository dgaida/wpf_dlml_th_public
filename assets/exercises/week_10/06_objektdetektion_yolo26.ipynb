## 2. Objektdetektion (Object Detection)

Objektdetektion identifiziert Objekte in einem Bild oder Video und zeichnet einen Begrenzungsrahmen (Bounding Box) um sie herum. Jede Box ist mit einer Klasse (z. B. "Auto", "Person") und einem Konfidenzwert (wie sicher sich das Modell ist) beschriftet.

Wir verwenden das kleinste Modell, **YOLO26n** (nano), welches für mobile Geräte und Echtzeitanwendungen optimiert ist.

```python
# Herunterladen eines Beispielvideos von TensorFlow (Traffic-Szene)
!wget -O /content/traffic_sample.mp4 https://raw.githubusercontent.com/tensorflow/models/master/research/object_detection/test_images/ducky.jpg
# Da das obige Verzeichnis meist Bilder enthält, nutzen wir eine zuverlässige alternative Quelle für ein kurzes MP4
!wget -O /content/sample_video.mp4 https://github.com/intel-iot-devkit/sample-videos/raw/master/bolt-detection.mp4
```

```python
from ultralytics import YOLO

# Modell laden
model_det = YOLO('yolo26n.pt')

# Inferenz auf dem neu heruntergeladenen Video
# Wir speichern das Ergebnis wieder unter /content/runs/detect/predict/
results = model_det.predict(source='/content/sample_video.mp4', conf=0.25, save=True, max_det=50)
```

## 3. Instanz-Segmentierung (Instance Segmentation)

Instanz-Segmentierung geht einen Schritt weiter als die Detektion: Anstatt nur einen Rahmen zu zeichnen, wird für jedes Objekt eine pixelgenaue Maske erstellt. So lässt sich die exakte Form des Objekts erfassen.

Hierfür nutzen wir das Modell **YOLO26n-seg**.

```python
# Lädt das vortrainierte YOLO26n-seg Modell für Segmentierung
model_seg = YOLO('yolo26n-seg.pt')

# Inferenz auf einem Beispielbild
results_seg = model_seg.predict(source='https://ultralytics.com/images/bus.jpg', save=True)
```

## Zusammenfassung

- **Objektdetektion**: Findet Objekte und markiert sie mit Bounding Boxes.
- **Instanz-Segmentierung**: Erstellt zusätzlich präzise Masken für jedes Objekt.
- **YOLO26**: Bietet verschiedene Modellgrößen (n, s, m, l, x) für unterschiedliche Anforderungen an Geschwindigkeit und Genauigkeit.
