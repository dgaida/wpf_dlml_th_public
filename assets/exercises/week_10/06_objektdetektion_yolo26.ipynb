## 2. Objektdetektion (Object Detection)  
  
Objektdetektion identifiziert Objekte in einem Bild und zeichnet einen Begrenzungsrahmen (Bounding Box) um sie herum.

```python
from ultralytics import YOLO  
from PIL import Image  
  
# Modell laden  
model = YOLO("yolo26n.pt")  # Lädt ein offizielles Modell  
  
# Inferenz auf einem Bild  
results = model("https://ultralytics.com/images/bus.jpg")  
  
# Ergebnisse visualisieren  
for r in results:  
    im_array = r.plot()  # BGR numpy array  
    im = Image.fromarray(im_array[..., ::-1])  # RGB PIL image  
    display(im)
```

## 3. Objekt Tracking  
  
Tracking verfolgt Objekte über mehrere Frames in einem Video hinweg und weist ihnen eindeutige IDs zu.

```python
model = YOLO("yolo26n.pt")  
  
# Tracking starten (BoT-SORT ist der Standard-Tracker)  
# 'show=True' öffnet ein Fenster (lokal), 'save=True' speichert das Ergebnisvideo  
results = model.track(source="https://www.youtube.com/watch?v=hmOuTQU-Z1E", show=False, save=True)
```

## 4. Segmentierung (Instance Segmentation)  
  
Instanz-Segmentierung erstellt für jedes Objekt eine pixelgenaue Maske.

```python
from ultralytics import YOLO  
from PIL import Image  
  
# Segmentierungs-Modell laden  
model = YOLO("yolo26n-seg.pt")  
  
# Inferenz auf einem Bild  
results = model("https://ultralytics.com/images/bus.jpg")  
  
# Ergebnisse visualisieren  
for r in results:  
    im_array = r.plot()  
    im = Image.fromarray(im_array[..., ::-1])  
    display(im)
```

## Zusammenfassung  
  
- **Objektdetektion**: Findet Objekte mit Bounding Boxes.  
- **Objekt Tracking**: Verfolgt Objekte über die Zeit.  
- **Instanz-Segmentierung**: Erstellt präzise Masken für jedes Objekt.
